```python
genre_popularity = df.groupby('track_genre')['popularity'].mean().sort_values(ascending=False).head(10)

plt.figure(figsize=(10,5))
genre_popularity.plot(kind='bar')
plt.xlabel('Genre')
plt.ylabel('Average Popularity')
plt.title('Top 10 Genres by Average Popularity')
plt.xticks(rotation=45)
plt.show()
```

From the visualizations, danceability and energy have some connection to popularity, but the trend isn't very strong. The genre chart shows clearer differences. Some genres have noticeably higher average popularity than others. I believe genre is more closely linked to popularity because some genres are currently more favored by listeners. Overall, popularity seems to depend on several factors rather than just one feature.
